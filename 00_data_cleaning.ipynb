```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense
from sklearn.ensemble import IsolationForest
from sklearn.model_selection import GroupKFold

# 1. Load your uploaded PhysioNet dataset
nights_df = pd.read_csv('bidsleep_nights.csv')
print(f"Loaded bidsleep_nights.csv: {len(nights_df)} nights | {nights_df['subject_id'].nunique()} people\n")

# 2. Teammate's exact baseline & anomaly planting logic
FEATURES = {
    "sleep_hr_median": "high",
    "sleep_hr_p5": "high",
    "total_sleep_h": "low",
    "wake_pct": "high",
    "n3_pct_of_sleep": "low",
    "rem_pct_of_sleep": "low",
}
HR_FEATURES = ["sleep_hr_median", "sleep_hr_p5"]
SLEEP_FEATURES = ["total_sleep_h", "wake_pct", "n3_pct_of_sleep", "rem_pct_of_sleep"]
TARGET_FALSE_ALARM = 0.05
SHIFTS_BPM = [3, 5, 8, 12]

def add_quality_flags(df):
    df = df.copy()
    reasons = pd.Series("", index=df.index)
    reasons[df["hr_per_sleep_hour"] < 500] += "low HR coverage; "
    reasons[df["hr_per_sleep_hour"] > 1000] += "about double the normal HR rate; "
    reasons[df["unknown_pct"] > 15] += "over 15% unknown sleep stage; "
    df["quality_issue"] = reasons.str.rstrip("; ")
    df["clean"] = df["quality_issue"] == ""
    return df

def loo_deviations(df):
    out = []
    for _, g in df.groupby("subject_id"):
        clean = g[g["clean"]]
        for idx, row in g.iterrows():
            others = clean.drop(index=idx, errors="ignore")
            dev = {f: row[f] - others[f].median() if len(others) >= 2 else np.nan for f in FEATURES}
            out.append({"idx": idx, "n_baseline": len(others), **{f"dev_{f}": v for f, v in dev.items()}})
    return pd.DataFrame(out).set_index("idx")

def pooled_spread(dev, people):
    sub = dev[dev["subject_id"].isin(people) & dev["clean"]]
    return {f: 1.4826 * np.nanmedian(np.abs(sub[f"dev_{f}"] - np.nanmedian(sub[f"dev_{f}"])))
            for f in FEATURES}

def z_scores(dev, spread):
    z = pd.DataFrame(index=dev.index)
    for f, direction in FEATURES.items():
        raw = dev[f"dev_{f}"] / spread[f]
        z[f] = raw if direction == "high" else -raw
    return z

def plant(df):
    base = df[df["clean"]]
    rows = []
    for idx, r in base.iterrows():
        for s in SHIFTS_BPM:
            p = r.copy()
            p["sleep_hr_median"] += s
            p["sleep_hr_p5"] += s
            rows.append({**p, "orig_idx": idx, "plant": f"HR +{s} bpm"})
        p = r.copy()
        lost = min(1.5, p["total_sleep_h"] - 2)
        p["total_sleep_h"] -= lost
        p["wake_pct"] += 15
        p["n3_pct_of_sleep"] = max(0, p["n3_pct_of_sleep"] - 8)
        rows.append({**p, "orig_idx": idx, "plant": "Disrupted sleep"})
    return pd.DataFrame(rows)

def deviations_for_planted(df, planted):
    out = []
    for _, r in planted.iterrows():
        others = df[(df["subject_id"] == r["subject_id"]) & df["clean"]].drop(index=r["orig_idx"])
        out.append({f"dev_{f}": r[f] - others[f].median() if len(others) >= 2 else np.nan for f in FEATURES})
    return pd.DataFrame(out, index=planted.index)

# 3. Your Autoencoder architecture (6 -> 4 -> 2 -> 4 -> 6)
def build_and_train_autoencoder(z_tr, seed=0):
    tf.keras.utils.set_random_seed(seed)
    X_tr = z_tr.to_numpy(dtype=np.float32)
    input_dim = X_tr.shape[1]

    inp = Input(shape=(input_dim,))
    enc = Dense(4, activation="relu")(inp)
    bot = Dense(2, activation="relu")(enc)
    dec = Dense(4, activation="relu")(bot)
    out = Dense(input_dim, activation="linear")(dec)

    ae = Model(inp, out)
    ae.compile(optimizer="adam", loss="mse")
    ae.fit(X_tr, X_tr, epochs=80, batch_size=16, verbose=0)
    return ae

def ae_mse_scores(ae, z_df):
    X = z_df.to_numpy(dtype=np.float32)
    recon = ae.predict(X, verbose=0)
    return np.mean(np.square(X - recon), axis=1)

# 4. Run GroupKFold Benchmark
def run_full_benchmark(nights, seed=0, n_splits=5):
    df = add_quality_flags(nights)
    dev = loo_deviations(df).join(df[["subject_id", "clean"]])
    ok = dev[[f"dev_{f}" for f in FEATURES]].notna().all(axis=1)

    planted = plant(df)
    pdev = deviations_for_planted(df, planted)
    pdev["subject_id"] = planted["subject_id"].to_numpy()
    pdev["plant"] = planted["plant"].to_numpy()
    pdev = pdev[pdev[[f"dev_{f}" for f in FEATURES]].notna().all(axis=1)]

    people = np.array(sorted(df["subject_id"].unique()))
    results = []

    for train_idx, test_idx in GroupKFold(n_splits=n_splits).split(people, groups=people):
        train_p, test_p = people[train_idx], people[test_idx]
        spread = pooled_spread(dev, train_p)

        z_tr = z_scores(dev[ok & dev["subject_id"].isin(train_p) & dev["clean"]], spread)
        z_te = z_scores(dev[ok & dev["subject_id"].isin(test_p) & dev["clean"]], spread)
        z_pl = z_scores(pdev[pdev["subject_id"].isin(test_p)], spread)
        plant_lbl = pdev.loc[z_pl.index, "plant"]

        # Detector A: Teammate's App Model
        half = TARGET_FALSE_ALARM / 2
        t_hr = np.quantile(z_tr[HR_FEATURES].max(axis=1), 1 - half)
        t_sl = np.quantile(z_tr[SLEEP_FEATURES].max(axis=1), 1 - half)
        flag = lambda z: (z[HR_FEATURES].max(axis=1) > t_hr) | (z[SLEEP_FEATURES].max(axis=1) > t_sl)

        results.append({"Detector": "z-score, separate HR & sleep (app)", "kind": "untouched",
                        "hits": int(flag(z_te).sum()), "n": len(z_te)})
        for lbl in plant_lbl.unique():
            m = (plant_lbl == lbl).to_numpy()
            results.append({"Detector": "z-score, separate HR & sleep (app)", "kind": lbl,
                            "hits": int(flag(z_pl[m]).sum()), "n": int(m.sum())})

        # Detector B: Isolation Forest
        iso = IsolationForest(n_estimators=300, random_state=seed).fit(z_tr)
        thr_if = np.quantile(-iso.score_samples(z_tr), 1 - TARGET_FALSE_ALARM)
        results.append({"Detector": "Isolation Forest (6 features)", "kind": "untouched",
                        "hits": int((-iso.score_samples(z_te) > thr_if).sum()), "n": len(z_te)})
        for lbl in plant_lbl.unique():
            m = (plant_lbl == lbl).to_numpy()
            results.append({"Detector": "Isolation Forest (6 features)", "kind": lbl,
                            "hits": int((-iso.score_samples(z_pl[m]) > thr_if).sum()), "n": int(m.sum())})

        # Detector C: Your Autoencoder
        ae = build_and_train_autoencoder(z_tr, seed=seed)
        thr_ae = np.quantile(ae_mse_scores(ae, z_tr), 1 - TARGET_FALSE_ALARM)
        results.append({"Detector": "Autoencoder (6-4-2-4-6)", "kind": "untouched",
                        "hits": int((ae_mse_scores(ae, z_te) > thr_ae).sum()), "n": len(z_te)})
        for lbl in plant_lbl.unique():
            m = (plant_lbl == lbl).to_numpy()
            results.append({"Detector": "Autoencoder (6-4-2-4-6)", "kind": lbl,
                            "hits": int((ae_mse_scores(ae, z_pl[m]) > thr_ae).sum()), "n": int(m.sum())})

    res_df = pd.DataFrame(results)
    agg = res_df.groupby(["Detector", "kind"])[["hits", "n"]].sum()
    agg["rate"] = agg["hits"] / agg["n"]

    rows = []
    for det, g in agg.groupby(level=0):
        g = g.droplevel(0)
        row = {"Detector": det, "False alarms (untouched)": f"{g.loc['untouched', 'rate']:.0%}"}
        for k in [f"HR +{s} bpm" for s in SHIFTS_BPM] + ["Disrupted sleep"]:
            row[k] = f"{g.loc[k, 'rate']:.0%}"
        rows.append(row)
    return pd.DataFrame(rows)

summary = run_full_benchmark(nights_df, seed=0, n_splits=5)
print("=== LEAVE-PEOPLE-OUT BENCHMARK RESULTS ===")
print(summary.to_string(index=False))
```

```python
import numpy as np
import pandas as pd
import tensorflow as tf
import joblib
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense

# Helper variables from the benchmark cell
FEATURES = {
    "sleep_hr_median": "high",
    "sleep_hr_p5": "high",
    "total_sleep_h": "low",
    "wake_pct": "high",
    "n3_pct_of_sleep": "low",
    "rem_pct_of_sleep": "low",
}
TARGET_FALSE_ALARM = 0.05

def add_quality_flags(df):
    df = df.copy()
    reasons = pd.Series("", index=df.index)
    reasons[df["hr_per_sleep_hour"] < 500] += "low HR coverage; "
    reasons[df["hr_per_sleep_hour"] > 1000] += "about double the normal HR rate; "
    reasons[df["unknown_pct"] > 15] += "over 15% unknown sleep stage; "
    df["quality_issue"] = reasons.str.rstrip("; ")
    df["clean"] = df["quality_issue"] == ""
    return df

def loo_deviations(df):
    out = []
    for _, g in df.groupby("subject_id"):
        clean = g[g["clean"]]
        for idx, row in g.iterrows():
            others = clean.drop(index=idx, errors="ignore")
            dev = {f: row[f] - others[f].median() if len(others) >= 2 else np.nan for f in FEATURES}
            out.append({"idx": idx, "n_baseline": len(others), **{f"dev_{f}": v for f, v in dev.items()}})
    return pd.DataFrame(out).set_index("idx")

def pooled_spread(dev, people):
    sub = dev[dev["subject_id"].isin(people) & dev["clean"]]
    return {f: 1.4826 * np.nanmedian(np.abs(sub[f"dev_{f}"] - np.nanmedian(sub[f"dev_{f}"])))
            for f in FEATURES}

def z_scores(dev, spread):
    z = pd.DataFrame(index=dev.index)
    for f, direction in FEATURES.items():
        raw = dev[f"dev_{f}"] / spread[f]
        z[f] = raw if direction == "high" else -raw
    return z

def build_and_train_autoencoder(z_tr, seed=0):
    tf.keras.utils.set_random_seed(seed)
    X_tr = z_tr.to_numpy(dtype=np.float32)
    input_dim = X_tr.shape[1]

    inp = Input(shape=(input_dim,))
    enc = Dense(4, activation="relu")(inp)
    bot = Dense(2, activation="relu")(enc)
    dec = Dense(4, activation="relu")(bot)
    out = Dense(input_dim, activation="linear")(dec)

    ae = Model(inp, out)
    ae.compile(optimizer="adam", loss="mse")
    ae.fit(X_tr, X_tr, epochs=80, batch_size=16, verbose=0)
    return ae

# 1. Re-load the data and compute Z-Scores on the whole clean dataset
nights_df = pd.read_csv('bidsleep_nights.csv')
df = add_quality_flags(nights_df)
dev = loo_deviations(df).join(df[["subject_id", "clean"]])
ok = dev[[f"dev_{f}" for f in FEATURES]].notna().all(axis=1)

# Calculate final population spread using all clean records
all_clean_subjects = df[df["clean"]]["subject_id"].unique()
final_spread = pooled_spread(dev, all_clean_subjects)

# Filter to untouched/clean sleep records
clean_dev_records = dev[ok & dev["clean"]]
z_full = z_scores(clean_dev_records, final_spread)

# 2. Build and train final Autoencoder model on all clean records
print("Training the final Autoencoder model on all clean data...")
final_model = build_and_train_autoencoder(z_full, seed=0)

# 3. Save the trained Keras model and the associated spread configuration
model_filename = 'final_autoencoder_model.keras'
final_model.save(model_filename)

config_filename = 'final_model_config.pkl'
joblib.dump({
    'spread': final_spread,
    'features': list(FEATURES.keys()),
    'target_alarm': TARGET_FALSE_ALARM
}, config_filename)

print(f"\nSuccess!")
print(f"- Model saved to: {model_filename}")
print(f"- Preprocessing config saved to: {config_filename}")
print("Refresh the Colab file sidebar folder to see and download them!")
```
